```python
arquivos_dia = os.listdir()
pdf = pdfplumber.open('/content/drive/MyDrive/Projeto_PDFtoJSON/receita2.pdf')
page = pdf.pages[0]

#corte da página para utilizar apenas a 1º via
primeira_via = page.crop((0 , 0 , page.width/2 , page.height))

#extrair os dados do PDF
text = primeira_via.extract_text()
text
```

```python
#estrutura para armazenar os dados
dados = {
    "CNS/CPF": None,
    "prescritor": None,
    "unid_saude": None,
    "medicamento": None,
    }

list(dados)

padrao_med = re.compile(
    r'\^s*\d+\.\s+(?P<nome>.+?)(:)
    )
```

```python
# A string que contém '20 comprimidos'
quantidade_str = text.split('\n')[13]

# Usar expressão regular para encontrar o primeiro grupo de dígitos
match = re.search(r'\d+', quantidade_str)

if match:
    numero_str = match.group(0)
    numero_float = float(numero_str)
    print(f"O número extraído é: {numero_float}")
    print(f"O tipo da variável é: {type(numero_float)}")
else:
    print("Nenhum número encontrado na string.")
```
