```python
import pandas as pd

# Check sheet names
excel_path = 'estoque_atenolol.xlsx'
xls = pd.ExcelFile(excel_path)
print("Sheet names:", xls.sheet_names)

df = pd.read_excel(excel_path)
print("Columns:", df.columns)
print("Head:")
print(df.head())
```

```python
import ast
import json

df_raw = pd.read_excel('estoque_atenolol.xlsx')

def parse_item(val):
    if isinstance(val, str):
        try:
            return json.loads(val.replace("'", '"')) # handle single quote JSON if needed
        except:
            return ast.literal_eval(val)
    return val

parsed_items = df_raw['items'].apply(parse_item)
df_items = pd.DataFrame(parsed_items.tolist())

print("Parsed items columns:", df_items.columns)
print("Shape:", df_items.shape)
print(df_items.head(10))
```

```python
# Save cleaned file
output_file = 'estoque_atenolol_organizado.xlsx'
df_items = df_items[['id', 'cnes', 'unidade', 'idProduto', 'produto', 'qtdEstoque', 'dataAtualizacao']]
df_items.to_excel(output_file, index=False)

print("Saved cleanly to", output_file)
print("Total rows:", len(df_items))
```
